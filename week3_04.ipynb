```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

nim = "244107020127"

def gamma_correction(image, gamma):
    inv_gamma = 1.0 / gamma if gamma > 0 else 1.0
    table = np.array([((i / 255.0) ** inv_gamma) * 255 for i in range(256)]).astype(np.uint8)
    return cv.LUT(image, table)

print(' Gamma Correction pada citra ')
print('------------------------------------')
try:
    gamma_input = float(input('Masukkan nilai Gamma: '))
except ValueError:
    print('Error, not a number')
    gamma_input = 0.5

path_citra = '/content/drive/MyDrive/mobile/KTP1a.jpg'
original = cv.imread(path_citra)

if original is None:
    print(f"File {path_citra} tidak ditemukan. Periksa path berkas Anda.")
else:
    img_gamma_input = gamma_correction(original, gamma_input)
    orig_rgb = cv.cvtColor(original, cv.COLOR_BGR2RGB)
    gamma_rgb = cv.cvtColor(img_gamma_input, cv.COLOR_BGR2RGB)

    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    fig.suptitle(f"Gamma Correction (Input = {gamma_input}) - NIM: {nim}", fontsize=13, fontweight='bold')

    axes[0].imshow(orig_rgb)
    axes[0].set_title(f"Citra Asli - {nim}")
    axes[0].axis('off')

    axes[1].imshow(gamma_rgb)
    axes[1].set_title(f"Gamma Correction (γ = {gamma_input}) - {nim}")
    axes[1].axis('off')

    plt.tight_layout()
    plt.show()

    gamma_candidates = [0.3, 0.5, 0.7, 1.0, 1.5, 2.0]

    fig_comp, axes_comp = plt.subplots(2, 3, figsize=(16, 9))
    fig_comp.suptitle(f"Eksplorasi Variasi Nilai Gamma pada KTM1a - NIM: {nim}", fontsize=14, fontweight='bold')
    axes_comp = axes_comp.flatten()

    for idx, g in enumerate(gamma_candidates):
        res = gamma_correction(original, g)
        res_rgb = cv.cvtColor(res, cv.COLOR_BGR2RGB)
        axes_comp[idx].imshow(res_rgb)
        axes_comp[idx].set_title(f"Gamma = {g} - {nim}")
        axes_comp[idx].axis('off')

    plt.tight_layout()
    plt.show()
```

```python
import cv2 as cv
import numpy as np
import glob
import math
import matplotlib.pyplot as plt
import pandas as pd

nim = "244107020127"

path_original = '/content/drive/MyDrive/mobile/galaxy.jpg'
original = cv.imread(path_original)

path_noises = '/content/drive/MyDrive/mobile/noises/*.jpg'
cv_img = []
for img in sorted(glob.glob(path_noises)):
    n = cv.imread(img)
    if n is not None:
        cv_img.append(n)

print(f"Total citra bernoise yang terbaca: {len(cv_img)}")

def hitung_psnr(img_asli, img_hasil):
    mse = np.mean((img_asli.astype(np.float64) - img_hasil.astype(np.float64)) ** 2)
    if mse == 0:
        return float('inf')
    max_pixel = 255.0
    psnr = 20 * math.log10(max_pixel / math.sqrt(mse))
    return psnr

jumlah_list = [10, 20, 40, 80, 100]
hasil_denoised = []
psnr_scores = []

for k in jumlah_list:
    subset = np.array(cv_img[:k], dtype=np.float64)
    avg_img = np.mean(subset, axis=0)
    avg_img_uint8 = np.clip(avg_img, 0, 255).astype(np.uint8)
    hasil_denoised.append(avg_img_uint8)

    psnr_val = hitung_psnr(original, avg_img_uint8)
    psnr_scores.append(round(psnr_val, 2))

fig, axes = plt.subplots(1, len(jumlah_list), figsize=(18, 4))
fig.suptitle(f"Hasil Average Denoising Berdasarkan Jumlah Citra - NIM: {nim}", fontsize=13, fontweight='bold')

for idx, k in enumerate(jumlah_list):
    res_rgb = cv.cvtColor(hasil_denoised[idx], cv.COLOR_BGR2RGB)
    axes[idx].imshow(res_rgb)
    axes[idx].set_title(f"N = {k} Citra\nPSNR: {psnr_scores[idx]} dB", fontsize=10)
    axes[idx].axis('off')

plt.tight_layout()
plt.show()
```

Hasil Analisis:

Operator NOT:Bekerja secara unari dengan menginversi setiap bit intensitas piksel ($s = 255 - A$). Piksel bernilai rendah (gelap) berubah menjadi tinggi (terang), dan sebaliknya, menghasilkan tampilan klise/negatif penuh tanpa melibatkan nilai piksel mask.

Operator OR:Memenuhi kaidah aljabar Boolean: $A \text{ OR } 255 = 255$ dan $A \text{ OR } 0 = A$. Pada area latar mask bernilai $255$ ($11111111_2$), seluruh bit dipaksa aktif menjadi $255$ (putih pekat). Pada area kotak wajah dan NIM bernilai $0$ ($00000000_2$), bit citra tidak berubah sama sekali, sehingga area yang justru ditargetkan untuk disamarkan tetap tampak asli sementara latarnya tertutup putih.

Operator AND:Memenuhi kaidah $A \text{ AND } 255 = A$ dan $A \text{ AND } 0 = 0$. Pada area latar mask bernilai $255$, bit citra dipertahankan apa adanya. Sebaliknya, pada kotak target berintensitas $0$, bit dikalikan dengan nol sehingga seluruh saluran warna teredam menjadi hitam pekat ($0$). Ini merupakan operator standar yang berhasil menyamarkan wajah dan NIM dengan kotak hitam.

Operator NAND:Merupakan operasi inversi dari hasil AND ($\text{NOT}(A \text{ AND } \text{Mask})$). Pada area kotak wajah dan NIM di mana hasil AND adalah $0$, negasinya membalik nilai tersebut menjadi $255$ (kotak putih penuh). Pada area latar di mana hasil AND adalah citra asli $A$, nilainya dibalik menjadi klise citra negatif ($255 - A$).

Operator XOR:Mengikuti logika eksklusif: bit bernilai $1$ jika kedua bit berbeda. Karena nilai $255$ secara biner adalah deretan bit $1$, operasi $A \oplus 255$ menghasilkan bit komplemen ($255 - A$, citra negatif pada area latar). Pada kotak wajah dan NIM berintensitas $0$, operasi $A \oplus 0 = A$ mempertahankan nilai asli piksel citra.


```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt

nim = "244107020127"

path_citra = '/content/drive/MyDrive/mobile/NIGHT1.jpg'
img_bgr = cv.imread(path_citra)

if img_bgr is None:
    print(f"File {path_citra} tidak ditemukan. Cek kembali penamaan dan path file.")
else:
    img_rgb = cv.cvtColor(img_bgr, cv.COLOR_BGR2RGB)

    gammas = [1.5, 2.5, 3.5, 5.0]

    fig, axes = plt.subplots(1, 5, figsize=(20, 4.5))
    fig.suptitle(f"Eksplorasi Gamma Correction Citra Malam - NIM: {nim}", fontsize=13, fontweight='bold')

    axes[0].imshow(img_rgb)
    axes[0].set_title(f"Original (Gelap) - {nim}")
    axes[0].axis('off')

    for idx, g in enumerate(gammas):
        table = np.array([((i / 255.0) ** (1.0 / g)) * 255 for i in range(256)]).astype(np.uint8)
        res = cv.LUT(img_rgb, table)

        axes[idx + 1].imshow(res)
        axes[idx + 1].set_title(f"Terang (Level {g}) - {nim}")
        axes[idx + 1].axis('off')

    plt.tight_layout()
    plt.show()
```

Alasan Pemilihan gamma = 3.5:

Citra NIGHT1.jpg mengalami extreme underexposure dengan mayoritas intensitas piksel kertas dan siluet berada di bawah nilai 30.   

Nilai eksponensial 1 / 3.5 =0.285 mendongkrak piksel-piksel bernilai sangat rendah tersebut naik ke batas $120 - 180$, sehingga lembaran kertas dan goresan objek di atasnya dapat terlihat jelas tanpa memerlukan penambahan linier konstan yang memutihkan langit secara menyeluruh.

Risiko Penggunaan gamma = 3.5:

Noise Amplification: Titik-titik derau sensor (color grain/chroma noise) pada bayangan kertas dan latar belakang malam yang semula tak kasatmata ikut terangkat dan terlihat bintik-bintik kasar.

Color Fading (Penurunan Saturasi): Sebagian rona hitam pekat pada langit malam berubah menjadi abu-abu pudar berkabut akibat regangan kurva yang sangat agresif.
