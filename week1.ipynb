```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
"https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
"https://external-preview.redd.it/hJNL3HXwmJXzzIdyUKdsVQerrBIe0KbO7d26zEXWidE.jpg?auto=webp&s=4a8bf4911d40dddb64ebc30505e704003297f414"]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
      image = io.imread(url) #read image
image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) #convert color to RGB
final_frame = cv.hconcat((image, image_2)) #concatenate image
image = io.imread(url) #read image (RGB)
image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image (RGB)
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) #convert image (RGB) to BGR for image_2
final_frame = cv.hconcat((cv.cvtColor(image, cv.COLOR_RGB2BGR), image_2)) # Convert image to BGR before concatenation f
cv2_imshow(final_frame) #show image
print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi," x ",lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)

```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
h, w, _ = image_2.shape
tengah_x = w // 2
cv.line(image_2, (tengah_x, 0), (tengah_x, h), (255, 255, 255), thickness=1)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
cv.line(image_2, (0, 0), (w, h), (255, 255, 255), thickness=1)
cv.line(image_2, (w, 0), (0, h), (255, 255, 255), thickness=1)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
h, w, _ = image_2.shape
tengah_y = h // 2
tengah_x = w // 2
panjang_garis = 200
x_start = tengah_x - (panjang_garis // 2)
x_end = tengah_x + (panjang_garis // 2)
cv.line(image_2, (x_start, tengah_y), (x_end, tengah_y), (255, 255, 255), thickness=1)
cv2_imshow(image_2)
```

```python
image_kotak = image.copy()
image_kotak[100:200, 100:200] = [255, 255, 255]
# Tampilkan bersebelahan
final_frame = cv.hconcat((image, image_kotak))
final_frame = cv.hconcat((cv.cvtColor(image, cv.COLOR_RGB2BGR), cv.cvtColor(image_kotak, cv.COLOR_RGB2BGR)))
cv2_imshow(final_frame)
```
